```python
from autogluon.tabular import TabularPredictor
import os
# 1. Defina o caminho para a subpasta exata onde o modelo foi salvo
diretorio_principal = '../AutogluonModels/' 
```

```python
# Lista para armazenar os dataframes de leaderboard de cada pasta
todos_os_leaderboards = []

print("Iniciando a extração dos tempos de treinamento...")

# 2. Varre todas as subpastas dentro do diretório principal
for nome_subpasta in os.listdir(diretorio_principal):
    caminho_completo = os.path.join(diretorio_principal, nome_subpasta)
    
    # Verifica se é uma pasta e se contém o arquivo essencial do AutoGluon (predictor.pkl)
    if os.path.isdir(caminho_completo) and 'predictor.pkl' in os.listdir(caminho_completo):
        print(f"Carregando dados da pasta: {nome_subpasta}...")
        
        try:
            # Carrega o predictor sem carregar os modelos pesados na memória
            predictor = TabularPredictor.load(caminho_completo)
            
            # Gera o leaderboard (ranking) que contém nativamente os tempos
            df_leaderboard = predictor.leaderboard(silent=True)
            
            # Adiciona uma coluna para você saber de qual pasta/experimento veio aquele modelo
            df_leaderboard['experimento_pasta'] = nome_subpasta
            
            # Adiciona o tempo TOTAL do .fit() daquela pasta como uma coluna extra (opcional)
            tempo_total_fit = predictor.info().get('time_fit_training', None)
            df_leaderboard['tempo_total_fit_pasta'] = tempo_total_fit
            
            # Guarda o resultado parcial
            todos_os_leaderboards.append(df_leaderboard)
            
        except Exception as e:
            print(f"Erro ao processar a pasta {nome_subpasta}: {e}")
```

```python
import pandas as pd

# 1. Carregue o CSV que geramos no passo anterior
# Lembre-se de verificar se o nome do arquivo e o separador (;) estão corretos
df = pd.read_csv('resultados_completos_autogluon.csv', sep=';')

# 2. Agrupe os dados pelo nome da pasta e faça a soma do tempo marginal
soma_tempos = df.groupby('experimento_pasta')['fit_time_marginal'].sum().reset_index()

# 3. Renomeie a coluna para deixar claro o que é
soma_tempos = soma_tempos.rename(columns={'fit_time_marginal': 'soma_tempo_modelos_seg'})


# Arredondando os valores para 2 casas decimais
soma_tempos = soma_tempos

# Exibe o resultado na tela
print("=== SOMA DOS TEMPOS DOS MODELOS POR PASTA ===")
print(soma_tempos)

# 5. (Opcional) Salve esse resumo em um novo CSV
soma_tempos.to_csv('resumo_tempos_por_pasta.csv', index=False, sep=';', encoding='utf-8-sig')
```

```python
import pandas as pd
# 3. Consolida todos os resultados em um único DataFrame e exporta para CSV
if todos_os_leaderboards:
    df_consolidado = pd.concat(todos_os_leaderboards, ignore_index=True)
    
    # Reorganiza as colunas para ficar mais fácil de analisar
    colunas_ordenadas = [
        'experimento_pasta', 'model', 'score_val', 
        'fit_time', 'pred_time_val', 'fit_time_marginal', 
        'pred_time_val_marginal', 'tempo_total_fit_pasta'
    ]
    # Garante que só vamos usar colunas que realmente existem no DataFrame
    colunas_existentes = [col for col in colunas_ordenadas if col in df_consolidado.columns]
    df_final = df_consolidado[colunas_existentes]
    
    # Salva o arquivo CSV
    nome_csv_saida = 'resultados_completos_autogluon.csv'
    df_final.to_csv(nome_csv_saida, index=False, sep=';', encoding='utf-8-sig')
    
    print(f"\nSucesso! O arquivo '{nome_csv_saida}' foi gerado com todos os tempos de treinamento.")
else:
    print("\nNenhuma pasta válida do AutoGluon foi encontrada ou processada.")
```

```python
import pandas as pd

# Arquivos de entrada e saída
arquivo_raw = 'C:/Users/Administrador/Downloads/machine_learning_2026_leandro/results/raw.csv'
arquivo_resumo = 'resumo_tempos_por_pasta.csv'
arquivo_saida = 'raw_corrigido_final.csv'

# ========================================================
# PASSO 1: Corrigir a quebra de linha do raw.csv
# ========================================================
linhas_corrigidas = []
with open(arquivo_raw, 'r', encoding='utf-8') as f:
    for linha in f:
        # Corrige o problema do 'task_id' grudado no meio da linha
        if 'task_id' in linha and not linha.startswith('task_id'):
            idx = linha.find('task_id')
            linhas_corrigidas.append(linha[:idx] + '\n')
            linhas_corrigidas.append(linha[idx:])
        else:
            linhas_corrigidas.append(linha)

# Salva um temporário limpo
with open('raw_temp.csv', 'w', encoding='utf-8') as f:
    f.writelines(linhas_corrigidas)

# ========================================================
# PASSO 2: Carregar os dados limpos
# ========================================================
df_raw = pd.read_csv('raw_temp.csv')
df_raw = df_raw[df_raw['task_id'] != 'task_id'].copy() # Remove cabeçalhos perdidos

# Forçar colunas numéricas
colunas_num = ['fit_time_s', 'predict_time_s', 'total_time_s']
for col in colunas_num:
    df_raw[col] = pd.to_numeric(df_raw[col], errors='coerce')

df_resumo = pd.read_csv(arquivo_resumo, sep=';')

# ========================================================
# PASSO 3: Pegar os "Últimos 60" de cada lado e cruzar
# ========================================================
# Identifica TODAS as linhas do AutoGluon no raw.csv
mask_ag = df_raw['model'].str.contains('autogluon', case=False, na=False)
todos_indices_ag = df_raw[mask_ag].index.tolist()

# Pega apenas os índices dos ÚLTIMOS 60 registros do AutoGluon
indices_ultimos_60_ag = todos_indices_ag[-60:]

# Pega as ÚLTIMAS 60 linhas do arquivo de pastas do AutoGluon
ultimas_60_pastas = df_resumo.tail(60).reset_index(drop=True)

print(f"Cruzando {len(indices_ultimos_60_ag)} registros com {len(ultimas_60_pastas)} pastas...")

# ========================================================
# PASSO 4: Atualizar os Tempos Sequencialmente
# ========================================================
# Substitui linha a linha na mesma ordem
for i, idx_raw in enumerate(indices_ultimos_60_ag):
    # Pega o tempo de treino correto da pasta correspondente
    novo_tempo_treino = ultimas_60_pastas.loc[i, 'soma_tempo_modelos_seg']
    
    # 1. Atualiza o fit_time_s
    df_raw.at[idx_raw, 'fit_time_s'] = novo_tempo_treino
    
    # 2. Recalcula o total_time_s (Treino Novo + Teste Original)
    tempo_teste = df_raw.at[idx_raw, 'predict_time_s']
    df_raw.at[idx_raw, 'total_time_s'] = novo_tempo_treino + tempo_teste

# ========================================================
# PASSO 5: Salvar o resultado
# ========================================================
df_raw.to_csv(arquivo_saida, index=False, encoding='utf-8')
print(f"\nSucesso absoluto! O arquivo '{arquivo_saida}' foi gerado.")
```
